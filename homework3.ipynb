```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso
from sklearn.linear_model import Ridge
from sklearn.linear_model import ElasticNet
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_validate

df = pd.DataFrame({
    "Age": [25, 30, 35, 40, 45, 50, 55, 60],
    "BMI": [22, 25, 27, 30, 32, 35, 37, 40],
    "Exercise Hours": [5, 4, 3, 2, 2, 1, 1, 0],
    "Medical Expense": [90, 120, 160, 220, 280, 340, 410, 500]
})
X = df[["Age", "BMI", "Exercise Hours"]]
y = df["Medical Expense"]

simple_model = LinearRegression()
simple_model.fit(X, y)

complex_model = Pipeline([
    ("poly", PolynomialFeatures(degree=4)),
    ("model", LinearRegression())
])
complex_model.fit(X, y)

new = pd.DataFrame({"Age": [42], "BMI": [31], "Exercise Hours": [2]})
print("SIMPLE MODEL PREDICT:", simple_model.predict(new))
print("COMPLEX MODEL PREDICT:", complex_model.predict(new))

model=Lasso(alpha=0.2, max_iter=100000)
model.fit(X, y)
model1 =Ridge(alpha=2.0)
model1.fit(X, y)
model2=ElasticNet(alpha=0.2, l1_ratio=0.6, max_iter=100000)
model2.fit(X, y)

print("Lasso coef:  ", model.coef_)
print("Ridge coef:  ", model1.coef_)
print("Elastic coef:", model2.coef_)

result = cross_validate(ElasticNet(alpha=0.2, l1_ratio=0.6, max_iter=100000),
                        X, y, cv=4, scoring="neg_mean_squared_error")
scores = result["test_score"]
print("CV scores:", scores)
print("Average score:", scores.mean())
```
