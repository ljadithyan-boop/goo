```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Lasso
from sklearn.linear_model import Ridge
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import cross_validate

df = pd.DataFrame({
    "apartment size": [600, 800, 1000, 1200, 1400, 1600, 1800, 2000],
    "price": [28, 38, 52, 68, 85, 105, 126, 150]
})
X = df[["apartment size"]]
y = df["price"]

simple_model = LinearRegression()
simple_model.fit(X, y)

complex_model = Pipeline([
    ("poly", PolynomialFeatures(degree=4)),
    ("model", LinearRegression())
])
complex_model.fit(X, y)

new = pd.DataFrame({"apartment size": [1700]})
print("SIMPLE MODEL PREDICT:", simple_model.predict(new))
print("COMPLEX MODEL PREDICT:", complex_model.predict(new))

model=Lasso(alpha=0.5)
model.fit(X,y)
print(model.coef_)

model=Ridge(alpha=0.1)
model.fit(X,y)
print(model.coef_)

model=ElasticNet(alpha=0.5,l1_ratio=0.5)
model.fit(X,y)
print(model.coef_)

result = cross_validate(Ridge(alpha=0.1), X, y, cv=5, scoring="neg_mean_squared_error")
scores = result["test_score"]
print("CV scores:", scores)
print("Average score:", scores.mean())

```
